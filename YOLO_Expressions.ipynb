# Import YOLO
Since you have already installed the Ultralytics Library you can easily import just the YOLO module.

```python
# Load YOLO library
from ultralytics import YOLO
```

## Begin Training the Model

```python
# Load a pre-trained YOLO model for object detection
model = YOLO('yolo11n.pt')

# Train the model
results = model.train(
    data='/content/dataset/data.yaml',
    epochs=100,
    patience=20,
    batch=32,
    imgsz=640,
    project='/content/drive/MyDrive/yolo_runs',
    name='yolo11n_run_final'
)
```

## Begin Testing the Model

```python
# Load the trained YOLO model.
model = YOLO("/content/drive/MyDrive/yolo_runs/yolo11n_run_final/weights/best.pt")

# Test the model
results = model.val(
    data='/content/dataset/data.yaml',
    split='test',
    project='/content/drive/MyDrive/yolo_runs',
    name='yolo11n_test_final'
)
```
